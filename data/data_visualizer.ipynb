This is just a quick file that lets you open up the json files in tabular form to get an easier view of the data

```python
import json
import pandas as pd

# call with a file path and "daily"
def visualize_file(filename):

    data = json.load(open(f"raw/{filename}"))
    df = pd.DataFrame(data["daily"])
    df["time"] = pd.to_datetime(df["time"])
    df.head()                              

    df.plot(x="time", y=["snow_depth_max", "snowfall_sum"], figsize=(12, 4))
    return df
```

```python
filename = "openmeteo/daily_best_match/grandvalira_base_2016-08-01_2016-12-31.json"

df = visualize_file(filename)
display(df)
```
